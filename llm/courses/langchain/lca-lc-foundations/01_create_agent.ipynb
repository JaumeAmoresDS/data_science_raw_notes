# 2. Tools



## Defining tool

- [video](https://academy.langchain.com/courses/take/foundation-introduction-to-langchain-python/lessons/71234851-lesson-2-tools)
- [notebook 1](https://github.com/langchain-ai/lca-lc-foundations/blob/main/notebooks/module-1/1.2_tools.ipynb)
- [notebook 2](https://github.com/langchain-ai/lca-lc-foundations/blob/main/notebooks/module-1/1.2_web_search.ipynb)


### Basic

- Decorator: `@tool`
- Name of function: square_root.
- Docstring explaining functionality and optionally when to use it.

### Alternative: put this info in decorator

- name of tool
- description (similar to docstring above)

### More hints: system prompt

- Indicate tools at disposal, relation with role of agent, and optionally when to use each tool.

### Invoke tool

```python
tool.invoke({first_arg: first_val, ...})
```

### Indicating to agent: 

```python
create_agent (
    ...,
    tools=[tool1, tool2,...]
)
```

# 4. Multimedia messages

## General

- Transform multimedia content to bytes using `io.BytesIO()`
- Transform from bytes to b64 using `base64`, encoding and decoding to utf-8
- HumanMessage content field has list of dicts, one per multimedia type, indicating type, encoding and, if not text, mime_type

## Image

- `FileUpload` (ipywidget)

## Audio

- `sounddevice.rec` for recording
- `scipy.io.wavfile.write` for converting to wav format
